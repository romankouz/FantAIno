```python
new_columns = []
for _, row in melondy_df.iterrows():
    spotify_data = get_spotify_album(row["artist"], row["album"])
    output = (
        total_tracks,
        num_available_markets,
        release_year,
        release_month,
        release_day,
        album_duration_in_s,
        explicit_proportion,
        featured_artists,
        num_features,
        track_names,
        artist_popularity,
    ) = process_spotify_album_data(spotify_data)
    new_columns.append(output)
    time.sleep(0.1)
```

```python
new_column_names = [
    "total_tracks",
    "num_available_markets",
    "release_year",
    "release_month",
    "release_day",
    "album_duration_in_s",
    "explicit_proportion",
    "featured_artists",
    "num_features",
    "track_names",
    "artist_popularity",
]

new_data_df = pd.DataFrame(new_columns, columns=new_column_names)
melondy_df[new_column_names] = new_data_df
```
